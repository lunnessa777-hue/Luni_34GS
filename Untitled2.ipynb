```python
import pandas as pd
import requests
from io import StringIO

headers = {"User-Agent": "Mozilla/5.0"}
seasons = ["1920", "2021", "2122", "2223", "2324", "2425", "2526", "2627"]

frames = []
for s in seasons:
    url = f"https://www.football-data.co.uk/mmz4281/{s}/T1.csv"
    try:
        r = requests.get(url, headers=headers, timeout=30)
        r.raise_for_status()
        d = pd.read_csv(StringIO(r.text))
        d["Season"] = s
        frames.append(d)
        print(s, "yüklendi:", len(d), "maç")
    except Exception as e:
        print(s, "yüklenemedi:", e)

df = pd.concat(frames, ignore_index=True)
print(df.shape)
print(df[["Season", "Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG"]].tail())
```

```python
# Gerekli sütunları seç
cols = ["Season", "Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG"]
data = df[cols].copy()

# Tarihi düzelt ve eksikleri at
data["Date"] = pd.to_datetime(data["Date"], dayfirst=True, errors="coerce")
data = data.dropna().sort_values("Date").reset_index(drop=True)
data["FTHG"] = data["FTHG"].astype(int)
data["FTAG"] = data["FTAG"].astype(int)

print("Temiz maç sayısı:", len(data))
print("Tarih aralığı:", data["Date"].min(), "->", data["Date"].max())

# Takım isimleri ve kaç maçta geçtikleri
teams = pd.concat([data["HomeTeam"], data["AwayTeam"]]).value_counts()
print("Takım sayısı:", len(teams))
print(teams.sort_index().to_string())

# Kapanış oranı sütunları var mı?
print([c for c in df.columns if "C" in c and c[-1] in "HDA"][:20])
```

```python
cols = ["Season", "Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "PSCH", "PSCD", "PSCA"]
data = df[cols].copy()

data["Date"] = pd.to_datetime(data["Date"], dayfirst=True, errors="coerce")
data = data.dropna(subset=["Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG"])
data = data.sort_values("Date").reset_index(drop=True)
data[["FTHG", "FTAG"]] = data[["FTHG", "FTAG"]].astype(int)

# Aynı kulübün iki ismini birleştir
fix = {"Erzurum BB": "Erzurumspor"}
data["HomeTeam"] = data["HomeTeam"].replace(fix)
data["AwayTeam"] = data["AwayTeam"].replace(fix)

print("Maç sayısı:", len(data))
print("Takım sayısı:", pd.concat([data["HomeTeam"], data["AwayTeam"]]).nunique())
print("Kapanış oranı olan maç sayısı (sezon bazında):")
print(data.groupby("Season")["PSCH"].apply(lambda s: s.notna().sum()))
```

```python
cols = ["Season", "Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG",
        "AvgCH", "AvgCD", "AvgCA"]
data = df[cols].copy()
data["Date"] = pd.to_datetime(data["Date"], dayfirst=True, errors="coerce")
data = data.dropna(subset=["Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG"])
data = data.sort_values("Date").reset_index(drop=True)
data[["FTHG", "FTAG"]] = data[["FTHG", "FTAG"]].astype(int)

fix = {"Erzurum BB": "Erzurumspor"}
data["HomeTeam"] = data["HomeTeam"].replace(fix)
data["AwayTeam"] = data["AwayTeam"].replace(fix)

# Sezon bazında dolu oran sayısı
print(data.groupby("Season")["AvgCH"].apply(lambda s: s.notna().sum()))

# Marjı çıkarıp olasılığa çevir
inv = 1 / data[["AvgCH", "AvgCD", "AvgCA"]]
total = inv.sum(axis=1)
data["pH"] = inv["AvgCH"] / total
data["pD"] = inv["AvgCD"] / total
data["pA"] = inv["AvgCA"] / total

print("Ortalama bahis şirketi marjı:", round((total - 1).mean() * 100, 2), "%")
print(data[["HomeTeam", "AwayTeam", "pH", "pD", "pA"]].tail())
```

```python
import penaltyblog as pb

train_seasons = ["1920", "2021", "2122", "2223", "2324", "2425"]
train = data[data["Season"].isin(train_seasons)]
test = data[~data["Season"].isin(train_seasons)]
print("Eğitim:", len(train), "maç | Test:", len(test), "maç")

weights = pb.models.dixon_coles_weights(train["Date"], 0.0018)

model = pb.models.DixonColesGoalModel(
    train["FTHG"], train["FTAG"],
    train["HomeTeam"], train["AwayTeam"],
    weights,
)
model.fit()
print(model)

# Testte olup eğitimde hiç görülmeyen takımlar
seen = set(train["HomeTeam"]) | set(train["AwayTeam"])
unseen = (set(test["HomeTeam"]) | set(test["AwayTeam"])) - seen
print("Eğitimde hiç görülmeyen takımlar:", unseen)
```

```python
import numpy as np

t = test[~test["HomeTeam"].isin(unseen) & ~test["AwayTeam"].isin(unseen)]
t = t.dropna(subset=["pH", "pD", "pA"]).copy()
print("Kıyaslanan maç sayısı:", len(t))

# Model tahminleri
P = []
for _, r in t.iterrows():
    p = model.predict(r["HomeTeam"], r["AwayTeam"])
    P.append(list(p.home_draw_away))
P = np.array(P)

# Piyasa tahminleri
M = t[["pH", "pD", "pA"]].values

# Basit referans: eğitim verisindeki sonuç oranları
y_train = np.where(train["FTHG"] > train["FTAG"], 0,
          np.where(train["FTHG"] == train["FTAG"], 1, 2))
freq = np.bincount(y_train, minlength=3) / len(y_train)
B = np.tile(freq, (len(t), 1))

# Gerçek sonuç
y = np.where(t["FTHG"] > t["FTAG"], 0,
    np.where(t["FTHG"] == t["FTAG"], 1, 2))
onehot = np.eye(3)[y]

def logloss(Q):
    return -np.mean(np.log(np.clip(Q[np.arange(len(y)), y], 1e-9, 1)))

def brier(Q):
    return np.mean(np.sum((Q - onehot) ** 2, axis=1))

print("Referans oranları (H, D, A):", freq.round(3))
print(f"Basit referans : log loss {logloss(B):.4f} | Brier {brier(B):.4f}")
print(f"Model          : log loss {logloss(P):.4f} | Brier {brier(P):.4f}")
print(f"Piyasa         : log loss {logloss(M):.4f} | Brier {brier(M):.4f}")
```

```python
test_start = pd.Timestamp("2025-08-01")
months = pd.date_range(test_start, data["Date"].max() + pd.offsets.MonthBegin(1), freq="MS")

P_all, M_all, B_all, y_all = [], [], [], []

for i in range(len(months) - 1):
    start, end = months[i], months[i + 1]
    tr = data[data["Date"] < start]
    blk = data[(data["Date"] >= start) & (data["Date"] < end)]
    blk = blk.dropna(subset=["pH", "pD", "pA"])
    if len(blk) == 0:
        continue

    w = pb.models.dixon_coles_weights(tr["Date"], 0.0018)
    m = pb.models.DixonColesGoalModel(
        tr["FTHG"], tr["FTAG"], tr["HomeTeam"], tr["AwayTeam"], w
    )
    m.fit()

    seen = set(tr["HomeTeam"]) | set(tr["AwayTeam"])
    blk = blk[blk["HomeTeam"].isin(seen) & blk["AwayTeam"].isin(seen)]

    ytr = np.where(tr["FTHG"] > tr["FTAG"], 0,
          np.where(tr["FTHG"] == tr["FTAG"], 1, 2))
    fr = np.bincount(ytr, minlength=3) / len(ytr)

    for _, r in blk.iterrows():
        p = m.predict(r["HomeTeam"], r["AwayTeam"])
        P_all.append(list(p.home_draw_away))
        M_all.append([r["pH"], r["pD"], r["pA"]])
        B_all.append(fr)
        y_all.append(0 if r["FTHG"] > r["FTAG"] else (1 if r["FTHG"] == r["FTAG"] else 2))

    print(start.strftime("%Y-%m"), "tamam,", len(blk), "maç")

P2, M2, B2, y2 = map(np.array, (P_all, M_all, B_all, y_all))
oh = np.eye(3)[y2]

def ll(Q): return -np.mean(np.log(np.clip(Q[np.arange(len(y2)), y2], 1e-9, 1)))
def br(Q): return np.mean(np.sum((Q - oh) ** 2, axis=1))

print("Kıyaslanan maç:", len(y2))
print(f"Basit referans : log loss {ll(B2):.4f} | Brier {br(B2):.4f}")
print(f"Model (aylık)  : log loss {ll(P2):.4f} | Brier {br(P2):.4f}")
print(f"Piyasa         : log loss {ll(M2):.4f} | Brier {br(M2):.4f}")
```

```python
val_start = pd.Timestamp("2024-08-01")
val_end = pd.Timestamp("2025-07-01")
months = pd.date_range(val_start, val_end, freq="MS")

def val_score(xi):
    P_, y_ = [], []
    for i in range(len(months) - 1):
        start, end = months[i], months[i + 1]
        tr = data[data["Date"] < start]
        blk = data[(data["Date"] >= start) & (data["Date"] < end)]
        if len(blk) == 0:
            continue
        w = pb.models.dixon_coles_weights(tr["Date"], xi)
        m = pb.models.DixonColesGoalModel(
            tr["FTHG"], tr["FTAG"], tr["HomeTeam"], tr["AwayTeam"], w
        )
        m.fit()
        seen = set(tr["HomeTeam"]) | set(tr["AwayTeam"])
        blk = blk[blk["HomeTeam"].isin(seen) & blk["AwayTeam"].isin(seen)]
        for _, r in blk.iterrows():
            p = m.predict(r["HomeTeam"], r["AwayTeam"])
            P_.append(list(p.home_draw_away))
            y_.append(0 if r["FTHG"] > r["FTAG"] else (1 if r["FTHG"] == r["FTAG"] else 2))
    P_, y_ = np.array(P_), np.array(y_)
    return -np.mean(np.log(np.clip(P_[np.arange(len(y_)), y_], 1e-9, 1))), len(y_)

for xi in [0.0, 0.0005, 0.0018, 0.003, 0.005]:
    score, n = val_score(xi)
    print(f"xi={xi}: log loss {score:.4f} ({n} maç)")
```

```python
XI = 0.005
test_start = pd.Timestamp("2025-08-01")
months = pd.date_range(test_start, data["Date"].max() + pd.offsets.MonthBegin(1), freq="MS")

P_all, M_all, y_all = [], [], []

for i in range(len(months) - 1):
    start, end = months[i], months[i + 1]
    tr = data[data["Date"] < start]
    blk = data[(data["Date"] >= start) & (data["Date"] < end)]
    blk = blk.dropna(subset=["pH", "pD", "pA"])
    if len(blk) == 0:
        continue

    w = pb.models.dixon_coles_weights(tr["Date"], XI)
    m = pb.models.DixonColesGoalModel(
        tr["FTHG"], tr["FTAG"], tr["HomeTeam"], tr["AwayTeam"], w
    )
    m.fit()

    seen = set(tr["HomeTeam"]) | set(tr["AwayTeam"])
    blk = blk[blk["HomeTeam"].isin(seen) & blk["AwayTeam"].isin(seen)]

    for _, r in blk.iterrows():
        p = m.predict(r["HomeTeam"], r["AwayTeam"])
        P_all.append(list(p.home_draw_away))
        M_all.append([r["pH"], r["pD"], r["pA"]])
        y_all.append(0 if r["FTHG"] > r["FTAG"] else (1 if r["FTHG"] == r["FTAG"] else 2))

P3, M3, y3 = map(np.array, (P_all, M_all, y_all))
oh = np.eye(3)[y3]
idx = np.arange(len(y3))

ll_model = -np.log(np.clip(P3[idx, y3], 1e-9, 1))
ll_mkt = -np.log(np.clip(M3[idx, y3], 1e-9, 1))
br_model = np.sum((P3 - oh) ** 2, axis=1)
br_mkt = np.sum((M3 - oh) ** 2, axis=1)

print("Kıyaslanan maç:", len(y3))
print(f"Model  : log loss {ll_model.mean():.4f} | Brier {br_model.mean():.4f}")
print(f"Piyasa : log loss {ll_mkt.mean():.4f} | Brier {br_mkt.mean():.4f}")

diff = ll_model - ll_mkt
se = diff.std(ddof=1) / np.sqrt(len(diff))
print(f"Log loss farkı (model - piyasa): {diff.mean():.4f} ± {se:.4f} (1 standart hata)")
```
